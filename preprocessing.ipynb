```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data/train.csv')
df.head()
```

```python
# Change all 'Year' columns to decades
df['Year'] = df['Year'].apply(lambda x: str(x)[:3] + '0')
df.head()
```

```python
# Add values to empty 'Finals' rows
df['Finals'] = df['Finals'].fillna("N/A")
df.head()
```

```python
# Change decimals in 'Ladder_Position_Prev' and 'Teams_Prev' to integers
print(df['Ladder_Position_Prev'])
print(df['Teams_Prev'])
```

```python
# Convert 'Ladder_Position_Prev' and 'Teams_Prev' to integers and handle NaN values
df['Ladder_Position_Prev'] = df['Ladder_Position_Prev'].apply(lambda x: int(x) if pd.notna(x) else 0)
df['Teams_Prev'] = df['Teams_Prev'].apply(lambda x: int(x) if pd.notna(x) else 0)

print(df['Ladder_Position_Prev'])
print(df['Teams_Prev'])
```
