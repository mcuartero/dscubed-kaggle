```python
import pandas as pd 
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("data/train.csv")

training_data = df.dropna(subset = ['Ladder_Position_Prev', 'Ladder_Position', 'Ladder_Position_Next'])
x = training_data[['Ladder_Position_Prev', 'Ladder_Position']]
y = training_data['Ladder_Position_Next']

model = LinearRegression()
model.fit(x,y)
print("Coefficient for Ladder_Position_Prev:", model.coef_[0])
print("Coefficient for Ladder_Position:", model.coef_[1])
print("Intercept:", model.intercept_)
print("R^2 Score:", model.score(x, y))

y_pred = model.predict(x)

plt.scatter(y, y_pred, alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')  # Diagonal reference line
plt.xlabel("Actual Ladder_Position_Next")
plt.ylabel("Predicted")
plt.title("Actual vs Predicted Ladder Position (Next Year)")
plt.grid(True)
plt.show()
```
